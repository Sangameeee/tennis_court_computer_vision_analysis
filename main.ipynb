```python
import os


repo_url = "https://github.com/Sangameeee/tennis_court_computer_vision_analysis.git"
folder_to_download = "scripts"


repo_name = repo_url.split('/')[-1].replace('.git', '')
if not os.path.exists(repo_name):
    print(f"Cloning {repo_url}...")
    !git clone {repo_url}
    print("Repository cloned successfully.")
else:
    print(f"Repository '{repo_name}' already exists. Skipping clone.")


source_folder_path = os.path.join(repo_name, folder_to_download)
if os.path.exists(source_folder_path):
    
    destination_path = './'
    print(f"Copying '{folder_to_download}' from '{repo_name}' to '{destination_path}'...")
    !cp -r {source_folder_path} {destination_path}
    print(f"'{folder_to_download}' copied successfully to {destination_path}.")
else:
    print(f"Error: Folder '{folder_to_download}' not found in the repository '{repo_name}'.")

```
